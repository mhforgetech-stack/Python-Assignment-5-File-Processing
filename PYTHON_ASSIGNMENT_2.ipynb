````python
def process_mess_file(filename):

    total = 0.0
    report_lines = []
    rejected_lines = []

    try:
        with open(filename, "r", encoding="utf-8") as file:

            for line_number, line in enumerate(file, start=1):

                line = line.strip()

                if not line:
                    continue

                try:
                    parts = line.split(",")

                    if len(parts) != 3:
                        raise ValueError("Expected 3 fields")

                    register_number = parts[0].strip()
                    name = parts[1].strip()
                    amount = float(parts[2].strip())

                    if not register_number or not name:
                        raise ValueError("Missing student details")

                    total += amount
                    report_lines.append(
                        f"{name} - {amount:.2f}"
                    )

                except (ValueError, IndexError) as error:
                    rejected_lines.append(
                        f"Line {line_number}: {line} - {error}"
                    )

        print("MESS BILL REPORT")
        print("\n".join(report_lines))
        print("Total:", f"{total:.2f}")

        print("\nREJECTED RECORDS")
        if rejected_lines:
            print("\n".join(rejected_lines))
        else:
            print("None")

    except FileNotFoundError:
        print("Error: Input file not found.")

    except PermissionError:
        print("Error: Permission denied.")

    except UnicodeDecodeError:
        print("Error: Unable to decode the file.")


process_mess_file("mess_oct.txt")
````

## Task 3.2 – Design a Custom Exception

**1. Define the Custom Exception**

```python
class InvalidRecord(Exception):
    """Raised when an attendance record is not in the expected format."""
```

**2. Where Should It Be Raised?**

The `InvalidRecord` exception should be raised inside the `parse_line()` function when:

- A record does not contain exactly three fields.
- The attendance status is neither `P` nor `A`.

**3. Where Should It Be Handled?**

The exception should be handled inside the `summarise()` function using `try` and `except InvalidRecord`.

When an invalid record is found, the program should record its line number and error message in the errors list and continue processing the remaining records.

**Conclusion:** A custom exception makes attendance-record validation clearer and allows damaged records to be handled without stopping the entire program.

````python

import os
import tempfile


class InvalidRecord(Exception):
    """Raised when an attendance line is not in the expected format."""


def parse_line(line):
    parts = line.strip().split(",")

    if len(parts) != 3:
        raise InvalidRecord(
            f"expected 3 fields, found {len(parts)}"
        )

    register_number, date, status = [part.strip() for part in parts]

    if status not in ("P", "A"):
        raise InvalidRecord(f"invalid status '{status}'")

    return register_number, date, status


def summarise(path):
    present, absent, errors = 0, 0, []

    try:
        with open(path, "r", encoding="utf-8") as file:
            for line_number, line in enumerate(file, start=1):
                try:
                    register_number, date, status = parse_line(line)

                except InvalidRecord as error:
                    errors.append((line_number, str(error)))

                else:
                    if status == "P":
                        present += 1
                    else:
                        absent += 1

    except FileNotFoundError as error:
        raise RuntimeError("Attendance file was not found.") from error

    except UnicodeDecodeError as error:
        raise RuntimeError("Attendance file has an encoding problem.") from error

    finally:
        # Final cleanup or logging can be performed here.
        pass

    return present, absent, errors


# Sample attendance data from the assignment
data = (
    "192311001,2026-10-08,P\n"
    "192311002,2026-10-08,A\n"
    "192311003,2026-10-08\n"
    "192311004,2026-10-08,L\n"
)

with tempfile.NamedTemporaryFile(
    "w", suffix=".csv", delete=False, encoding="utf-8"
) as tmp:
    tmp.write(data)
    filename = tmp.name

try:
    print(summarise(filename))

except RuntimeError as error:
    print(error)

finally:
    if os.path.exists(filename):
        os.remove(filename)


# Test a missing file
try:
    summarise("missing_attendance.csv")

except RuntimeError as error:
    print("Missing file test:", error)
    print("Original cause:", type(error.__cause__).__name__)
````

# Task 4 – File and Exception Forensics

## Task 4.1 – Locate the Defects

**1. Why are earlier days' records lost?**

The program opens `returns.log` using write mode (`"w"`). Write mode truncates the existing file whenever it is opened. Therefore, each new day overwrites the previous day's records.

**2. Why are the books joined together?**

The program uses `log.write(b)` without adding a newline character (`"\n"`). As a result, consecutive book names are written directly next to each other.

**3. What happens if `write()` fails?**

If `write()` raises an exception, the `log.close()` statement may never execute. The file can remain open until Python eventually closes it during cleanup. This can cause resource leaks.

**4. Additional defect**

The input file is opened using `open(day_file).read()` without a `with` statement. The input file is also not explicitly closed.

**5. Incomplete exception handling**

The program catches only `FileNotFoundError`. Other errors, such as `PermissionError`, are not handled or recorded in an error log.

**Conclusion:** Incorrect file modes, missing newline characters, unsafe file handling and incomplete exception handling cause the program's failures.

### Alternative Concept and Complexity Analysis

Manual file handling can use `try/finally` to close files. If implemented correctly, it also ensures files are closed when errors occur. Both approaches have the same time complexity, O(n), where n is the number of books processed, and the same space complexity.

With `with` statements, no files remain open after their blocks finish. With correctly implemented `try/finally`, no files remain open either. Without proper cleanup, manually opened files may remain open.

````python

%%writefile attendance_single.py

# 1. Reading attendance files
def read_attendance(filename):
    records = []

    with open(filename, "r", encoding="utf-8") as file:
        for line in file:
            parts = line.strip().split(",")

            if len(parts) == 3 and parts[2] in ("P", "A"):
                records.append(parts)

    return records


# 2. Computing attendance statistics
def calculate_percentage(records):
    total = len(records)
    present = sum(1 for record in records if record[2] == "P")

    if total == 0:
        return 0

    return (present / total) * 100


# 3. Writing attendance reports
def write_report(records, percentage, filename):
    with open(filename, "w", encoding="utf-8") as file:
        file.write("ATTENDANCE REPORT\n")

        for record in records:
            file.write(",".join(record) + "\n")

        file.write(f"Attendance Percentage: {percentage:.2f}%\n")


# Main program
if __name__ == "__main__":
    sample_data = (
        "192311001,2026-10-08,P\n"
        "192311002,2026-10-08,A\n"
        "192311003,2026-10-08,P\n"
    )

    with open("sample_attendance.txt", "w", encoding="utf-8") as file:
        file.write(sample_data)

    records = read_attendance("sample_attendance.txt")
    percentage = calculate_percentage(records)
    write_report(records, percentage, "single_report.txt")

    print("Attendance:", f"{percentage:.2f}%")
    print("Report created successfully.")
````

### Stage 3: Identify Critical Design Decisions

**1. Duplicate Detection**

A Python set will store register numbers that have already been processed. Before accepting a registration, the program checks whether its register number exists in the set. This prevents duplicate registrations.

**2. Email Validation**

String methods will be used to perform basic email validation:
- Check that the email contains `@`.
- Check that the email contains a dot (`.`) after `@`.
- Reject empty email addresses and addresses containing spaces.

**3. Exception Handling**

The program will use specific exception handling:
- `FileNotFoundError` – handles missing department files.
- `PermissionError` – handles files that cannot be accessed.
- `UnicodeDecodeError` – handles encoding problems.
- `ValueError` – handles invalid record data.

**4. Error Logging**

Invalid records, duplicate registrations and file-related errors will be recorded in `error_log.txt`.

**5. File Management**

The program will use `with open()` to automatically close files after processing. It will continue processing the remaining files even when one file is missing or contains invalid data.

**Conclusion:**

Using sets, string validation, specific exceptions and error logging makes the registration system reliable, organised and easy to maintain.

### Stage 4: Design the Package

**Package Structure**

The College Fest Registration Desk will be organised into the following Python files:

```text
college_fest/
│
├── __init__.py
├── reader.py
├── validator.py
├── processor.py
├── report.py
└── main.py
```

**Module Responsibilities**

1. `reader.py`
   - Reads registration records from department files.
   - Handles file-opening operations.

2. `validator.py`
   - Checks the number of fields.
   - Validates register numbers, names, departments and email addresses.

3. `processor.py`
   - Detects duplicate registrations using a set.
   - Organises valid students by event.
   - Records invalid and duplicate entries.

4. `report.py`
   - Creates the clean master list.
   - Generates event-wise participant files.
   - Creates the summary report and error log.

5. `main.py`
   - Coordinates all modules.
   - Processes the five department files.
   - Ensures processing continues when an error occurs.

**Pseudocode**

```text
START

Create an empty set for register numbers
Create an empty master registration list
Create event-wise participant lists

FOR each department file:
    TRY to open the file
        Read each line
        Split the line into five fields
        Validate the record

        IF record is invalid:
            Write error to error log
            CONTINUE

        IF register number already exists:
            Write duplicate error to log
            CONTINUE

        Add register number to the set
        Add record to master list
        Add student to selected event lists

    EXCEPT file-related errors:
        Write error to log
        Continue with the next department file

Generate clean master list
Generate event-wise participant files
Generate summary report
Generate error log

STOP
```

**Conclusion:**

A modular package separates reading, validation, processing and reporting. This improves code reusability, readability, testing and maintenance.

````python

from pathlib import Path
import re

# Create folders
Path("college_fest").mkdir(exist_ok=True)
Path("input").mkdir(exist_ok=True)
Path("output").mkdir(exist_ok=True)

# Create package files
Path("college_fest/__init__.py").write_text("")

Path("college_fest/validator.py").write_text("""
def validate_record(line):
    fields = [item.strip() for item in line.strip().split(",")]

    if len(fields) != 5:
        raise ValueError("Record must contain exactly 5 fields")

    reg, name, department, email, events = fields

    if not all([reg, name, department, email, events]):
        raise ValueError("One or more fields are empty")

    if " " in email or email.count("@") != 1:
        raise ValueError("Invalid email address")

    local, domain = email.split("@")

    if not local or "." not in domain or domain.startswith(".") or domain.endswith("."):
        raise ValueError("Invalid email address")

    return {
        "register_number": reg,
        "name": name,
        "department": department,
        "email": email,
        "events": [e.strip() for e in events.split(";") if e.strip()]
    }
""")

Path("college_fest/reader.py").write_text("""
def read_department_file(filename):
    with open(filename, "r", encoding="utf-8") as file:
        for line_number, line in enumerate(file, start=1):
            yield line_number, line
""")

Path("college_fest/processor.py").write_text("""
from .reader import read_department_file
from .validator import validate_record

def process_files(files):
    master = []
    seen = set()
    event_lists = {}
    errors = []

    for filename in files:
        try:
            for line_number, line in read_department_file(filename):
                try:
                    student = validate_record(line)

                    reg = student["register_number"]

                    if reg in seen:
                        raise ValueError("Duplicate registration number")

                    seen.add(reg)
                    master.append(student)

                    for event in student["events"]:
                        event_lists.setdefault(event, []).append(student)

                except ValueError as error:
                    errors.append(
                        f"{filename}, line {line_number}: {error}"
                    )

        except FileNotFoundError:
            errors.append(f"{filename}: Department file not found")

        except PermissionError:
            errors.append(f"{filename}: Permission denied")

        except UnicodeDecodeError:
            errors.append(f"{filename}: Encoding error")

    return master, event_lists, errors
""")

Path("college_fest/report.py").write_text("""
from pathlib import Path

def generate_reports(master, event_lists, errors):
    output = Path("output")
    output.mkdir(exist_ok=True)

    with open(output / "master_list.txt", "w", encoding="utf-8") as file:
        for student in master:
            file.write(
                f"{student['register_number']},"
                f"{student['name']},"
                f"{student['department']},"
                f"{student['email']}\\n"
            )

    for event, students in event_lists.items():
        safe_name = "".join(
            c if c.isalnum() or c in "_-" else "_"
            for c in event
        )
        with open(output / f"{safe_name}_participants.txt",
                  "w", encoding="utf-8") as file:
            for student in students:
                file.write(
                    f"{student['register_number']},"
                    f"{student['name']},"
                    f"{student['department']}\\n"
                )

    with open(output / "summary_report.txt", "w", encoding="utf-8") as file:
        file.write(f"Total valid registrations: {len(master)}\\n")
        file.write(f"Total events: {len(event_lists)}\\n")
        for event, students in event_lists.items():
            file.write(f"{event}: {len(students)} participants\\n")

    with open(output / "error_log.txt", "w", encoding="utf-8") as file:
        if errors:
            file.write("\\n".join(errors))
        else:
            file.write("No errors found.")
""")

Path("college_fest/main.py").write_text("""
from .processor import process_files
from .report import generate_reports

def main():
    department_files = [
        "input/cse.txt",
        "input/it.txt",
        "input/aiml.txt",
        "input/ece.txt",
        "input/mech.txt"
    ]

    master, events, errors = process_files(department_files)
    generate_reports(master, events, errors)

    print("COLLEGE FEST REGISTRATION RESULTS")
    print("Valid registrations:", len(master))
    print("Events generated:", len(events))
    print("Errors recorded:", len(errors))
    print("Reports saved in the output folder.")

if __name__ == "__main__":
    main()
""")

# Sample department input files
Path("input/cse.txt").write_text(
    "192426001,Arun,CSE,arun@gmail.com,Music;Dance\n"
    "192426002,Divya,CSE,divya@gmail.com,Drama\n"
    "192426003,Kumar,CSE,kumar@gmail.com,Music\n"
)

Path("input/it.txt").write_text(
    "192426004,Meena,IT,meena@gmail.com,Dance;Drama\n"
    "192426001,Arun,CSE,arun@gmail.com,Music\n"
)

Path("input/aiml.txt").write_text(
    "192426005,Ravi,AIML,raviemail.com,Music\n"
    "192426006,Sara,AIML,sara@gmail.com,Drama\n"
)

Path("input/ece.txt").write_text(
    "192426007,John,ECE,john@gmail.com\n"
    "192426008,Asha,ECE,asha@gmail.com,Dance\n"
)

# mech.txt is intentionally not created to test missing-file handling.

print("Package, modules and sample input files created successfully!")
````
